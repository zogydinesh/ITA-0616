```python
data = """Sky,AirTemp,Humidity,Wind,Water,Forecast,EnjoySport
Sunny,Warm,Normal,Strong,Warm,Same,Yes
Sunny,Warm,High,Strong,Warm,Same,Yes
Rainy,Cold,High,Strong,Warm,Change,No
Sunny,Warm,High,Strong,Cool,Change,Yes"""

with open('enjoysport.csv', 'w') as f:
    f.write(data)
```

```python
import csv

with open('enjoysport.csv') as f:
    rows = list(csv.reader(f))
    header, data = rows[0], rows[1:]
    n = len(header) - 1

    h = ['0'] * n
    print("Initial h:", h, "\n")

    for step, row in enumerate(data, 1):
        inst, label = row[:-1], row[-1].strip().lower()

        if label == 'yes':
            for i in range(n):
                if h[i] == '0':
                    h[i] = inst[i]
                elif h[i] != inst[i]:
                    h[i] = '?'

        print(f"Example {step}: {inst} -> {row[-1]}")
        print("  h =", h, "\n")

    print("Most specific hypothesis (FIND-S):", h)
```
